# Physics Solution Colab One-Click

Drive에 `physics_solution.zip`과 `open.zip`만 올려두고 `런타임 > 모두 실행`

```python
from google.colab import drive
drive.mount('/content/drive')

import shutil
from pathlib import Path

# ── 여기만 수정 ──
DRIVE_BASE = Path('/content/drive/MyDrive/daCon/version_0.05')
PROJECT_ZIP = DRIVE_BASE / 'physics_solution.zip'
DATASET_ZIP = DRIVE_BASE / 'open.zip'

assert PROJECT_ZIP.exists(), f'Not found: {PROJECT_ZIP}'
assert DATASET_ZIP.exists(), f'Not found: {DATASET_ZIP}'

# 프로젝트 코드를 /content에 추출
if Path('/content/physics_solution').exists():
    shutil.rmtree('/content/physics_solution')
shutil.unpack_archive(str(PROJECT_ZIP), '/content')
print('Done:', Path('/content/physics_solution'))
```

```python
%cd /content/physics_solution
# ── v0.05 추천 파라미터 ──────────────────────────────────────────────────────
# batch-size: 32 (16→32), image-size: 336, backbone-lr: 2e-5,
# tta-passes: 8, num-workers: 4, weight-decay: 5e-5
# ─────────────────────────────────────────────────────────────────────────────
!bash run_colab_oneclick.sh \
    --drive-zip-path "{DATASET_ZIP}" \
    --batch-size 32 \
    --num-workers 4 \
    --tta-passes 8 \
    --image-size 336 \
    --backbone-lr 2e-5 \
    --weight-decay 5e-5
```

```python
runs = sorted(Path('/content/drive/MyDrive/physics_solution_outputs').glob('run_*'))
if runs:
    print('Latest output:', runs[-1])
    print('Submission:', runs[-1] / 'submission.csv')
```
